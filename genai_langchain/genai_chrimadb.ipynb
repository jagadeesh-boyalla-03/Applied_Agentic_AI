# Using ChromaDB with Sentence Transformer 

```python
# using specifc  emmbeding model for building Chromadb vector db

!pip install -U sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer
```

```python
# picking the embedding model

embedding_model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
texts = [
        "Python is most popular programing language",
        "Maching learning is a branch of AI"
]
embeddings = embedding_model.encode(texts)
print(embeddings)
```

```python
print(len(embeddings[0]))
```

```python
import chromadb
client = chromadb.Client()
```

```python
collection4 = client.get_or_create_collection(name = "documents")
```

```python
collection4.add(
    documents = [
        "Python is most pouplar programing language",
        "machine learing is  a branch of of AI"
    ],
    embeddings = embeddings, # we are inserting the sentense tranformer base embedding
    ids = ["doc1","doc2"]
)
```

```python
res5  = collection4.query(
    query_texts = ["Ai is future"],
    n_results = 1
)
res5
```

# ChromaDB with Langchain

```python
!pip install langchain-chroma
```

```python
!pip install langchain-huggingface
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
```

```python
# embedding model

embeddings = HuggingFaceEmbeddings(
    model_name = "sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
vectorstore = Chroma(
    collection_name = "my_documents",
    embedding_function = embeddings,
    persist_directory = "./Lang_Chroma_db"
)
```

```python
from langchain_core.documents import Document

documents = [
    Document(
        page_content = "python is most popular language",
        metadat = {"tpoic":"python"}
    ),
     Document(
        page_content = "machine learning part of ai",
        metadat = {"tpoic":"Ml"}
     )
]
```

```python
vectorstore.add_documents(documents)
```

```python
res6 = vectorstore.similarity_search("Ai is growing very fast",k=1)
```

# Chromadb with ollama embedding mode and ollama LLM

```python
import ollama
import chromadb
```

```python
client = chromadb.PersistentClient(path = "./ollama_chromdb")
```

```python
collection5 = client.get_or_create_collection(name = "documents")
```

```python
documents = [
    "Python is very simple programing language",
    "MYSQL is a RDBMS",
    "Machine learning is branch of AI",
    "Deep learning is based upon Neural Network"
]

ids = ["doc1","doc2","doc3","doc4"]
```

```python
!ollama pull nomic-embed-text
```

```python
# using ollama emmbeding model
embeddings = []
for d in documents:
    response = ollama.embed(model = "nomic-embed-text",input = d)
    embeddings.append(response["embeddings"][0])
```

```python
collection5.add(
    documents = documents,
    ids = ids,
    embeddings = embeddings 
)
```

```python
Query= "Ai Powerful tool"

response = ollama.embed(model = "nomic-embed-text",input = Query )
query_embedding = response["embeddings"][0]
```

```python
res7 = collection5.query(query_embeddings = [query_embedding],n_results=2)
res7
```
