```python
# promot template : used to customize promot 
# e.g : write poem on {topic}

from langchain_core.prompts import PromptTemplate
prompt = PromptTemplate.from_template("Write a poem of {topic}")

llm_obj = ChatGroq(
    model = "openai/gpt-oss-20b",
    temperature = 0.2,
)
chain = prompt | llm_obj 

user_ip = input("Enter a topic")

response = chain.invoke({"topic": user_ip})
print(response.content)
```

```python
# getting  output from llm in Json formate using langchain
# rasing exception in langchain 

# client -->  lagchain --> LLM


#client -->>lang chain -->llM
from langchain_core.prompts import PromptTemplate
prompt = PromptTemplate.from_template("Write a peom on {topic}")
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.exceptions import OutputParserException

```

```python
from langchain_groq import ChatGroq
parser = JsonOutputParser()
llm_obj = ChatGroq(model_name = "openai/gpt-oss-20b")
chain = prompt | llm_obj | parser
user_topic  = input("Enter your topic")
try:
    response = chain.invoke({"topic" : user_topic})
    print(response)
except OutputParserException as e:
    print(f"Output is not in json formate : {e}")
```
